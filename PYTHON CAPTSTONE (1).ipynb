```python
import numpy as np
import pandas as pd
```

```python
# -------------------------
# PROJECT DATAFRAME (Table 1)
# -------------------------

data_project = np.array([
    ["A001", "Project 1", 1002000, "Finished"],
    ["A002", "Project 2", 2000000, "Ongoing"],
    ["A003", "Project 3", 4500000, "Finished"],
    ["A004", "Project 4", 5500000, "Ongoing"],
    ["A005", "Project 5", np.nan,  "Finished"],   # Missing value
    ["A002", "Project 6", 680000, "Failed"],
    ["A005", "Project 7", 400000, "Finished"],
    ["A003", "Project 8", 350000, "Failed"],
    ["A001", "Project 9", np.nan,  "Ongoing"],    # Missing value
    ["A003", "Project 10", 300000, "Finished"],
    ["A001", "Project 11", 2000000, "Failed"],
    ["A004", "Project 12", 1000000, "Ongoing"],
    ["A004", "Project 13", 3000000, "Finished"],
    ["A005", "Project 14", 200000, "Finished"]
])

#columns of the data set
columns_project = ["ID", "Project", "Cost", "Status"]


df_project = pd.DataFrame(data_project, columns=columns_project)

# Save CSV as required
df_project.to_csv("Project_DataFrame.csv", index=False)


#To print the data set please uncomment the below print statement
# print(df_project)
```

```python
# -------------------------
# EMPLOYEE DATAFRAME (Table 2)
# -------------------------


employee_data = np.array([
    ["A001", "John Alter",        "M", "Paris",   25],
    ["A002", "Alice Luxumberg",   "F", "London",  27],
    ["A003", "Tom Sabestine",     "M", "Berlin",  29],
    ["A004", "Nina Adgra",        "F", "Newyork", 31],
    ["A005", "Amy Johny",         "F", "Madrid",  30]
])


#columns of the data set
columns_employee = ["ID", "Name", "Gender", "City", "Age"]

df_employee = pd.DataFrame(employee_data, columns=columns_employee)

# Save CSV exactly as required
df_employee.to_csv("Employee_DataFrame.csv", index=False)

#To print the data set please uncomment the below print statement
# print(df_employee)
```

```python
# -------------------------
# SENIORITY LEVEL DATAFRAME (Table 3)
# -------------------------

data_seniority = np.array([
    ["A001", 2],
    ["A002", 2],
    ["A003", 3],
    ["A004", 2],
    ["A005", 3]
], dtype=object)

columns_seniority = ["ID", "Designation Level"]

df_seniority = pd.DataFrame(data_seniority, columns=columns_seniority)

#pd.to_numeric is for assurance that designation level is in numeric
df_seniority["Designation Level"] = pd.to_numeric(df_seniority["Designation Level"], errors="coerce").astype("Int64")

# Save CSV exactly as instructed
df_seniority.to_csv("Seniority_Level_DataFrame.csv", index=False)

#To print the data set please uncomment the below print statement
# print(df_seniority)
```

```python
# --------------------------------------------------
'''Task 2 :- The cost column in the dataframe “Project” has some missing values. Your task is to
compute these missing values. Replace the missing values by running average. You
should use the “For” loop for this task.'''
# --------------------------------------------------


# To load CSV 
df_project = pd.read_csv("Project_DataFrame.csv")

# To Ensure that the column cost is numeric datatype
df_project["Cost"] = pd.to_numeric(df_project["Cost"], errors="coerce")

running_sum = 0.0
running_count = 0

for i in range(len(df_project)):
    if pd.isna(df_project.loc[i, "Cost"]):
        df_project.loc[i, "Cost"] = running_sum / running_count
    else:
        running_sum += df_project.loc[i, "Cost"]
        running_count += 1

# Save updated result for the next tasks
df_project.to_csv("Project_DataFrame_Task2.csv", index=False)

#To print the Updated dataset please uncomment the below statement
# df_project
```

```python
# --------------------------------------------------
'''Task 3 :- Split the name column in the Employee dataframe into two new columns “First Name”,
and “LastName” and remove the older “name” column..'''
# --------------------------------------------------

# To load CSV 
df_employee = pd.read_csv("Employee_DataFrame.csv")


df_employee["First Name"] = df_employee["Name"].str.split().str[0]
df_employee["Last Name"] = df_employee["Name"].str.split().str[1]

#to drop the name column according to given task
df_employee = df_employee.drop(columns=["Name"])


# Save updated result for the next tasks
df_employee.to_csv("Employee_DataFrame_Task3.csv", index=False)

#To print the Updated dataset please uncomment the below statement
# df_employee
```

```python
# --------------------------------------------------
'''Task 4 :- Join all three dataframes in one single dataframe. Name it Final'''
# --------------------------------------------------

#To load CSVs

'''Note:- I am using taks wise save file for project and employee dataframe but for seniority dataframe it is the same as created first'''

df_project = pd.read_csv("Project_DataFrame_Task2.csv")
df_employee = pd.read_csv("Employee_DataFrame_Task3.csv")
df_seniority = pd.read_csv("Seniority_Level_DataFrame.csv")

#Merging All three dataset into One "FINAL" 
Final = df_project.merge(df_employee, on="ID").merge(df_seniority, on="ID")

#Saving Merge files in one "FINAL" named file
Final.to_csv("Final_Task4.csv", index=False)

#To print the Updated "FINAL" dataset please uncomment the below statement
# Final
```

```python
# --------------------------------------------------
'''Task 5 :- Add a new bonus column in the Final dataframe. Give a 5% bonus concerning project
cost only to employees who have finished the projects.'''
# --------------------------------------------------

#Loading "FINAL" named dataset
Final = pd.read_csv("Final_Task4.csv")

Final["Bonus"] = 0.0

for i in range(len(Final)):
    if Final.loc[i, "Status"] == "Finished":
        Final.loc[i, "Bonus"] = Final.loc[i, "Cost"] * 0.05
        
#Saving Updated "FINAL" Dataset after adding bonus column
Final.to_csv("Final_Task5.csv", index=False)

#To print the Updated "FINAL" dataset please uncomment the below statement
# Final
```

```python
# --------------------------------------------------
'''Task 6 :-Demote the designation level by 1, whose projects have status “fail”. Delete the
employees record whose designation level is above 4.'''
# --------------------------------------------------

#To Load Last updated "FINAL" dataset
Final = pd.read_csv("Final_Task5.csv")

for i in range(len(Final)):
    if Final.loc[i, "Status"] == "Failed":
        Final.loc[i, "Designation Level"] -= 1
        
#To Delete the Employees record whose designation level is above 4 >>>>>>>
Final = Final[Final["Designation Level"] <= 4]

##Saving Updated "FINAL" Dataset after Task 6
Final.to_csv("Final_Task6.csv", index=False)

#To print the Updated "FINAL" dataset please uncomment the below statement
# Final
```

```python
# --------------------------------------------------
'''Task 7 :-Add “Mr.” and “Mrs.” to the first name column and drop the gender column.'''
# --------------------------------------------------

#To load Last updated "FINAL" dataframe
Final = pd.read_csv("Final_Task6.csv")

for i in range(len(Final)):
    if Final.loc[i, "Gender"] == "M":
        Final.loc[i, "First Name"] = "Mr. " + Final.loc[i, "First Name"]
    else:
        Final.loc[i, "First Name"] = "Mrs. " + Final.loc[i, "First Name"]

#TO dropping the Gender column
Final = Final.drop(columns=["Gender"])

#Saving Updated "FINAL" Dataset after Task 7
Final.to_csv("Final_Task7.csv", index=False)

#To print the Updated "FINAL" dataset please uncomment the below statement
# Final
```

```python
# --------------------------------------------------
'''Task 8 :-Promote designation level by 1 for the employees whose age is more than 29 years using
IF condition.'''
# --------------------------------------------------


#To load last updated "FINAL" datafrome
Final = pd.read_csv("Final_Task7.csv")

for i in range(len(Final)):
    if Final.loc[i, "Age"] > 29:
        Final.loc[i, "Designation Level"] += 1
        

#Saving Updated "FINAL" Dataset after Task 8
Final.to_csv("Final_Task8.csv", index=False)

#To print the Updated "FINAL" dataset please uncomment the below statement
# Final
```

```python
# --------------------------------------------------
'''Task 9 :-Add the cost of all projects for each Employee and save it in new dataframe
“TotalProjCost” with three columns ID, First Name, and Total cost'''
# --------------------------------------------------

#To load last updated "FINAL" datafrome
Final = pd.read_csv("Final_Task8.csv")

#for Grouping and adding the cost for each employee and creating new dataframe
TotalProjCost = Final.groupby(["ID", "First Name"])["Cost"].sum().reset_index()


#Renaming the columns according to tasks
TotalProjCost.rename(columns={"Cost": "Total cost"}, inplace=True)


#Saving the CSVs
TotalProjCost.to_csv("TotalProjCost_Task9.csv", index=False)


#To print the "TotalProjCost" dataset please uncomment the below statement
# TotalProjCost
```

```python
# --------------------------------------------------
'''Task 10 :-Print all the employee details whose city name contains the letter “o” in it'''
# --------------------------------------------------

#To load CSV
'''Note:- I am using task 8 csv file for performing this task because tasks 9 csv has grouped and total cost'''

Final = pd.read_csv("Final_Task8.csv")

#According to the task printing those employees whose city contains letter "O"------
Output = Final[Final["City"].str.contains("o", case=False)]

#To print the "TotalProjCost" dataset please uncomment the below statement
# Output
```
